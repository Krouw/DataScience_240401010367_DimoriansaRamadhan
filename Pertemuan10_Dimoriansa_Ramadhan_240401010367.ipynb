```python
# Langkah 1 : Muat dan Eksplorasi Data
# Baca dataset, periksa tipe data, dan hitung proporsi kelas churn untuk memastikan dataset bersifat imbalanced.

import pandas as pd
df = pd.read_csv("/content/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(df.shape)
print(df["Churn"].value_counts(normalize=True))
```

```python
# Langkah 2 : Preprocessing
# Lakukan encoding pada fitur kategorikal menggunakan pd.get_dummies() atau LabelEncoder, lalu bagi data menjadi latih/uji secara stratified untuk menjaga proporsi kelas:

from sklearn.model_selection import train_test_split
# TODO: encoding fitur kategorikal (pd.get_dummies)
df = pd.get_dummies(df, drop_first=True)
# TODO: pisahkan X (fitur) dan y (target = Churn)
X = df.drop("Churn_Yes", axis=1)
y = df["Churn_Yes"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
# Langkah 3 : Latih Model
# Bangun RandomForestClassifier dengan class_weight="balanced" untuk menangani imbalance:

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
n_estimators=300, class_weight="balanced", random_state=42)
rf.fit(X_tr, y_tr)
```

```python
# Langkah 4 : Evaluasi
# Hitung precision, recall, F1-score, dan ROC-AUC dengan fokus pada kelas churn (kelas 1):

from sklearn.metrics import classification_report, roc_auc_score

# Prediksi kelas
y_pred = rf.predict(X_te)

# Prediksi probabilitas kelas churn (kelas 1)
y_prob = rf.predict_proba(X_te)[:, 1]

# Menampilkan Precision, Recall, F1-score
print(classification_report(y_te, y_pred))

# Menghitung ROC-AUC
roc_auc = roc_auc_score(y_te, y_prob)
print("ROC-AUC:", roc_auc)
```

```python
# Langkah 5 : Prediksi Probabilitas dan Simpulkan
# Gunakan predict_proba untuk menghasilkan probabilitas churn tiap pelanggan

# Menghitung probabilitas churn (kelas 1)
churn_prob = rf.predict_proba(X_te)[:, 1]

# Menampilkan 10 probabilitas pertama
print("10 Probabilitas Churn Pertama:")
print(churn_prob[:10])
```

Kesimpulan mengenai temuan utama dari model :

Berdasarkan hasil prediksi, model Random Forest dapat memberikan probabilitas kemungkinan setiap pelanggan melakukan churn. Nilai evaluasi yang berasal dari precision, recall, F1-score, dan ROC-AUC menunjukkan bahwa model mampu membedakan pelanggan yang berpotensi churn dengan cukup baik. Informasi probabilitas tersebut dapat digunakan perusahaan untuk mengidentifikasi pelanggan yang memiliki risiko churn lebih tinggi sehingga dapat diberikan penawaran atau layanan khusus.
